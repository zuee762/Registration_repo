```python
df['brand'] = df['brand'].astype('string')
df[['model','milage','fuel_type']] = df[['model','milage','fuel_type']].astype('string')
df[['engine','transmission','accident','clean_title','price']] = df[['engine','transmission','accident','clean_title','price']].astype('string')
```

```python
df.drop(columns = ['ext_col','int_col'], inplace= True)
```

```python
df['brand'] = df['brand'].str.strip().str.lower()
df['fuel_type'] = df['fuel_type'].str.strip().str.lower()
df['engine'] = df['engine'].str.strip().str.lower()
df['transmission'] = df['transmission'].str.strip().str.lower()
df['accident'] = df['accident'].str.strip().str.lower()
df['clean_title'] = df['clean_title'].str.strip().str.lower()
```

```python
df['milage'] = df['milage'].str.replace(',','',regex=False).str.replace('mi','',regex=False).str.replace('.','',regex=False).str.strip()
df['price'] = df['price'].str.replace('$','',regex=False).str.replace(',','',regex=False).str.replace('.','',regex=False).str.strip()
df[['price','milage']] = df[['price','milage']].astype('int')
```

```python
df = df[df['model_year'] >= 2000]
```

```python
df.duplicated().sum()
```

```python
df['fuel_type'].isnull().sum()
df['fuel_type'].value_counts()
df['fuel_type'].unique()
df[df['fuel_type'].isnull()]
df['fuel_type'] = df['fuel_type'].fillna('unknown')
```

```python
df['accident'].isnull().sum()
df['accident'].unique()
df['accident'].value_counts()
df[df['accident'].isnull()]
df['accident'] = df['accident'].fillna('unknown')
df['accident'] = df['accident'].replace({
    'at least 1 accident or damage reported': 'Yes',
    'none reported': 'No',
    'unknown': 'unknown'
})
```

```python
df['clean_title'].isnull().sum()
df['clean_title'].unique()
df['clean_title'].value_counts()
df[df['clean_title'].isnull()]
df['clean_title'] = df['clean_title'].fillna('unknown')
```

```python
df['brand'].value_counts()
df['brand'].unique()
df['model'].value_counts()
df['model'].unique()
```

```python
df['accident'].value_counts()
df['accident'] = df['accident'].replace({
    'Yes': '1',
    'No': '0',
    'unknown': '0'
})
df['accident'] = df['accident'].astype('int')
```

```python
df['clean_title'].value_counts()

df['clean_title'] = df['clean_title'].replace({
    'yes': '1',
    'unknown': '0'
})
df['clean_title'] = df['clean_title'].astype('int')
```

```python
df.drop(columns = ['model'], inplace= True)
```
